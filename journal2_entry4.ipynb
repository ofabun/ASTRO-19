```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'Times New Roman'
plt.rc('font', size=15)
```

```python
# pulling from distribution - I chose the beta distribution
x = np.random.beta(0.1, 100, 1000)
x
```

```python
# plotting the beta dist in a hist
plt.hist(x, bins=100, color='cornflowerblue', ec='white')
plt.xlabel('x'); plt.ylabel('beta'); plt.title('Beta Distribution')

plt.xlim(0,0.01)
```
